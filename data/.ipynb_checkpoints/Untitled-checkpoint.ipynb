```python
def make_dirs(name_dir, song_dict, k):
    for i in range(9):
        game['categories'][k]['questions'][i]['audio'] = f'{name_dir}/{song_dict[i]}.mp3'
        game['categories'][k]['questions'][i]['answer'] = song_dict[i].replace('_', ' ')
```

```python
make_dirs('mmm', mmm, 0)
make_dirs('motematic', motematic, 1)
make_dirs('nerv', nerv, 7)
make_dirs('pilots', pilots, 8)
make_dirs('id', ids, 9)
game
```

```python
game['final']['audio'] = f'Gotye_-_Somebody_That_I_Used_To_Know.mp3'
game['final']['answer'] = 'Gotye_&_Kimbra_-_Somebody_That_I_Used_To_Know'.replace('_', ' ')
game
```
