```python
def make_dirs(name_dir, song_dict, name_cat, k):
    for i in range(9):
        game['categories'][k]['questions'][i]['audio'] = f'{name_dir}/{song_dict[i]}.mp3'
        game['categories'][k]['questions'][i]['answer'] = song_dict[i].replace('_', ' ')
        game['categories'][k]['title'] = name_cat
```

```python
make_dirs('mmm', mmm, 'ммм', 0)
make_dirs('motematic', motematic, 'мотематика', 1)
make_dirs('girls', girls, 'я на стороне дефчонок', 2)
make_dirs('rock', rock, 'еее роцк', 3)
make_dirs('rus', rus, 'я - русский', 4)
make_dirs('nerv', nerv, 'потреплем нервы', 5)
make_dirs('pilots', pilots, 'потому что мы пилоты', 6)
make_dirs('id', ids, 'не представляйте динозавры', 7)
make_dirs('alt', alt, 'альтернатива есть!', 8)
game
```
