```python
import json
```

```python
with open('questions.json', 'r', encoding='utf-8') as file:
    game = json.load(file)
game
```

```python
mmm = dict(zip(range(7), ['Alec_Benjamin - If_I_Killed_Someone_for_You', 'Dean_Lewis - Hurtless', 
                          'Isak_Danielson - Broken', 'James_Bay_Julia_Michaels - Peer_Pressure', 
                          'Lewis_Capaldi - Heavenly_Kind_Of_State_Of_Mind', 
                          'Noah_Kahan - Stick_Season', 'Passenger - Let_Her_Go']))
mmm
```

```python
for i in range(7):
    game['categories'][0]['questions'][i]['audio'] = f'mmm/{mmm[i]}.mp3'
    game['categories'][0]['questions'][i]['answer'] = mmm[i]
```

```python
with open('questions.json', 'w', encoding='utf-8') as file:
    json.dump(game, file)
```
