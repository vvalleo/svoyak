```python
import json
```

```python
with open('questions.json', 'r', encoding='utf-8') as file:
    game = json.load(file)
game
```

```python
mmm = dict(zip(range(8), ['Alec_Benjamin - If_I_Killed_Someone_for_You', 'Dean_Lewis - Small_Disasters', 
                          'Isak_Danielson - Broken', 'James_Bay_Julia_Michaels - Peer_Pressure', 
                          'Lewis_Capaldi - Heavenly_Kind_Of_State_Of_Mind', 'Damiano_David - Next_Summer',
                          'Noah_Kahan - Stick_Season', 'Passenger - Let_Her_Go']))
mmm
```

```python
motematic = dict(zip(range(8), ['One_Direction_-_Story_of_My_Life', '3_Doors_Down_-_Let_Me_Be_Myself', 
                          'Three_Days_Grace_-_Last_to_Know', 'Maroon_5_-_Memories', 
                          '30_Seconds_to_Mars_-_Up_In_The_Air', 'Sum_41_-_War',
                          'blink-182_-_All_The_Small_Things', 'Thousand_Foot_Krutch_-_Shook']))
motematic
```

```python
def make_dirs(name_dir, song_dict, k):
    for i in range(8):
        game['categories'][k]['questions'][i]['audio'] = f'{name_dir}/{song_dict[i]}.mp3'
        game['categories'][k]['questions'][i]['answer'] = song_dict[i].replace('_', ' ')
```

```python
make_dirs('mmm', mmm, 0)
make_dirs('motematic', motematic, 1)
```
